## 1) My Lane as an ML Task (Type)

* **Task Type:** Supervised Binary Classification & Ranking / Scoring
* **Domain Context:** Automated Resume Screening & Candidate Skill Alignment
* **Primary Objective:** Predict candidate suitability scores and classify whether a candidate profile matches high-priority job requirements based on extracted semantic features.
* **Downstream Action Supported:** Automatically flags top-tier candidate profiles for technical interviews and filters out non-qualifying submissions, reducing manual screening time for HR teams.
*

## 2) Target or Proxy

* **Target Variable:** `is_shortlisted` (Binary: `1` for Shortlisted, `0` for Rejected)
* **Proxy Metric / Variable:** `semantic_match_score` (Continuous value between `0.0` and `1.0`, calculated using embedding cosine similarity between candidate resume features and job descriptions).
* **Target Justification:** Direct historical hiring decisions (`is_shortlisted`) can be noisy or biased. Using a combined proxy metric (`semantic_match_score` validated against recruiter approval) provides a stable label for training models.
*

## 3) Success Metric

* **Primary Metric:** **Precision@K** (e.g., Precision@10) and **ROC-AUC**
* **Business / System Success Definition:**
  * At least **85%** of candidates recommended in the top 10 positions must be rated as qualified by domain recruiters (High Precision@10).
  * Achieve an overall **ROC-AUC $\ge$ 0.82** while maintaining fairness constraints (maintaining demographic parity / disparate impact ratio $\ge$ 0.80).

```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Simulate starter dataset slice representing raw candidate profiles and job descriptions
num_samples = 100

data = {
    "candidate_id": [f"CAND_{1000 + i}" for i in range(num_samples)],
    "years_experience": np.random.randint(1, 12, size=num_samples),
    "skills_count": np.random.randint(2, 15, size=num_samples),
    "education_level": np.random.choice(
        ["Bachelors", "Masters", "PhD"], size=num_samples, p=[0.6, 0.35, 0.05]
    ),
    "raw_resume_text": [
        "Python, Data Analysis, SQL, Machine Learning experience..."
        for _ in range(num_samples)
    ],
    "semantic_match_score": np.round(np.random.uniform(0.40, 0.98, size=num_samples), 3),
}

# Create starter DataFrame
df_starter = pd.DataFrame(data)

# Derive binary target column based on proxy score threshold
df_starter["is_shortlisted"] = (df_starter["semantic_match_score"] >= 0.75).astype(int)

print(f"Starter Dataset Shape: {df_starter.shape}")
df_starter.head(5)
```

## 5) Why ML Beats a Fixed Rule Here

1. **Contextual & Semantic Variability:** Fixed heuristic rules (e.g., `IF years_experience >= 3 AND 'Python' IN skills THEN shortlist`) fail when candidates use alternative phrasing (e.g., "PyData ecosystem", "Pandas & Scikit-Learn", or "Backend Data Pipelines").
2. **Non-Linear Feature Interaction:** Success is rarely determined by a single feature threshold; a candidate with 2 years of experience and exceptional domain project work might outperform a candidate with 5 years of unrelated experience. ML captures complex non-linear combinations that fixed `IF-ELSE` statements cannot represent without becoming unmaintainable.
3. **Adaptability & Continuous Improvement:** Fixed rules remain static and quickly become outdated as job requirements evolve. An ML model can be retrained on new recruiter feedback to continuously adjust scoring weights.
