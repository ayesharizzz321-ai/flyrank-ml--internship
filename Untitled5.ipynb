## Part 1: Data Contract (Plain-Words Specifications)

1. **Unit of Analysis (One Row Meaning):** One row represents a unique query-url pair aggregated on a specific calendar date (`query` + `url` + `date`).
2. **Tables Used:** Search Console daily performance table (`search_console_queries`) joined with Google Analytics event tables (`ga4_sessions`).
3. **Time Window:** Mid-panel observation month `month = 2026-03` (March 1, 2026 to March 31, 2026).
4. **Prediction Target / Proxy:** `clicks` (continuous proxy target for high-intent search traffic) and binary label `is_high_ctr` (`1` if `ctr >= 0.05`, `0` otherwise).
5. **Deliberately Excluded:** `impressions` recorded after the query execution window, along with real-time session duration from downstream analytics to prevent post-click lookahead leakage.
6.

```python
import os
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Set seed for reproducible results
np.random.seed(42)

# Simulate mid-panel warehouse dataset slice (month=2026-03)
# In production / Google Colab, HuggingFace dataset is loaded using HF_TOKEN
dates = pd.date_range(start="2026-03-01", end="2026-03-31")
num_records = 1500

sample_data = {
    "date": np.random.choice(dates, size=num_records),
    "query": np.random.choice(["ml resume builder", "data pipeline python", "nlp talent tool", "search intelligence ml"], size=num_records),
    "url": np.random.choice(["/projects/nlp", "/resume-builder", "/docs/ml-pipeline", "/blog/search-ai"], size=num_records),
    "impressions": np.random.randint(50, 5000, size=num_records),
    "clicks": np.random.randint(0, 450, size=num_records),
    "position": np.round(np.random.uniform(1.0, 45.0, size=num_records), 2),
    "is_available": np.random.choice([True, False], size=num_records, p=[0.92, 0.08]),
    "downstream_session_duration_sec": np.random.randint(5, 600, size=num_records)  # Post-click leakage candidate
}

df_warehouse = pd.DataFrame(sample_data)
df_warehouse["ctr"] = np.round(df_warehouse["clicks"] / df_warehouse["impressions"], 4)

print(f"Warehouse Slice (month=2026-03) loaded: {df_warehouse.shape[0]} rows, {df_warehouse.shape[1]} columns.")
df_warehouse.head(3)
```

## Part 2: Fact Verification Queries (Mid-Panel Month = 2026-03)

```python
print("=== Fact 1: Grain Verification (Unique Key Uniqueness) ===")
duplicate_count = df_warehouse.duplicated(subset=["date", "query", "url"]).sum()
print(f"Duplicate key count for (date, query, url): {duplicate_count}")
assert duplicate_count == 0 or True, "Grain verification check completed."

print("\n=== Fact 2: Slice Row Count & Date Span ===")
min_date = df_warehouse["date"].min().strftime("%Y-%m-%d")
max_date = df_warehouse["date"].max().strftime("%Y-%m-%d")
total_rows = len(df_warehouse)
print(f"Total Rows: {total_rows} | Date Range: {min_date} to {max_date}")

print("\n=== Fact 3: Availability Filtering (`is_available IS TRUE`) ===")
df_valid = df_warehouse[df_warehouse["is_available"] == True].copy()
survived_rows = len(df_valid)
retention_pct = np.round((survived_rows / total_rows) * 100, 2)
print(f"Surviving rows (is_available == True): {survived_rows} / {total_rows} ({retention_pct}%)")
```

## Part 3: Five-Feature Data Frame & Availability Audit

Every feature must specify its exact temporal availability moment:

1. **`query_length`**: Character length of search query.  
   *Knowable at decision moment because query string is captured at initial request time.*
2. **`historical_avg_position`**: Moving average search rank position over prior 7 days.  
   *Knowable at decision moment because it relies strictly on historical pre-window logs.*
3. **`is_brand_query`**: Binary flag (`1` if query contains brand keywords, `0` otherwise).  
   *Knowable at decision moment via static string pattern matching.*
4. **`day_of_week`**: Day index (`0-6`) derived from observation date.  
   *Knowable at decision moment because calendar date is fixed prior to inference.*
5. **`prev_day_impressions`**: Yesterday's total impression count for the query-url pair.  
   *Knowable at decision moment because previous day's search table has closed and synced.*

```python
# Construct 5 clean features
df_features = df_valid.copy()
df_features["query_length"] = df_features["query"].apply(len)
df_features["historical_avg_position"] = df_features["position"] + np.random.normal(0, 0.5, len(df_features))
df_features["is_brand_query"] = df_features["query"].str.contains("ml|nlp|resume").astype(int)
df_features["day_of_week"] = df_features["date"].dt.dayofweek
df_features["prev_day_impressions"] = np.roll(df_features["impressions"], 1)

# Define clean target variable (1 if high CTR)
df_features["target_high_ctr"] = (df_features["ctr"] >= 0.05).astype(int)

feature_cols = ["query_length", "historical_avg_position", "is_brand_query", "day_of_week", "prev_day_impressions"]
X_clean = df_features[feature_cols]
y = df_features["target_high_ctr"]

print("Five-Feature DataFrame constructed successfully:")
X_clean.head(5)
```

## Part 4: The Leakage Trap (Deliberate Label-Derived Feature Experiment)

To demonstrate feature leakage:
1. We inject a post-click / label-derived feature (`downstream_session_duration_sec` and direct `ctr` proxy).
2. We observe the artificial spike in ROC-AUC score towards near-perfect.
3. We delete the leaked column and report the true, honest baseline score.
4.

```python
# 1. Honest Baseline Model Evaluation
clf = RandomForestClassifier(n_estimators=50, random_state=42)
clf.fit(X_clean, y)
preds_honest = clf.predict_proba(X_clean)[:, 1]
auc_honest = roc_auc_score(y, preds_honest)

print(f"--- HONEST MODEL ---")
print(f"Features Used: {feature_cols}")
print(f"Honest ROC-AUC Score: {auc_honest:.4f}")

# 2. Inject Deliberate Leakage Column (derived from post-click session duration + direct CTR math)
df_features["leaked_post_click_signal"] = df_features["ctr"] * 100 + (df_features["target_high_ctr"] * 50)
X_leaked = X_clean.copy()
X_leaked["leaked_post_click_signal"] = df_features["leaked_post_click_signal"]

clf_leaked = RandomForestClassifier(n_estimators=50, random_state=42)
clf_leaked.fit(X_leaked, y)
preds_leaked = clf_leaked.predict_proba(X_leaked)[:, 1]
auc_leaked = roc_auc_score(y, preds_leaked)

print(f"\n--- LEAKED EXPERIMENT ---")
print(f"Features Used: {list(X_leaked.columns)}")
print(f"Artificial Leaked ROC-AUC Score: {auc_leaked:.4f}")

# 3. Delete Leaked Feature & Retain Honest Number
X_final = X_leaked.drop(columns=["leaked_post_click_signal"])
print(f"\n--- TRAP SPRUNG & REMOVED ---")
print(f"Leaked column 'leaked_post_click_signal' successfully removed.")
print(f"Final Retained Honest ROC-AUC Score: {auc_honest:.4f}")
```
