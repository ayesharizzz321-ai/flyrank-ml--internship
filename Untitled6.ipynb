```python
import pandas as pd
import numpy as np

# Create a dummy DataFrame with the expected columns
data = {
    'query_id': np.arange(100),
    'position': np.random.randint(1, 10, 100),
    'ctr': np.random.rand(100) * 0.1,
    'impressions': np.random.randint(100, 1000, 100),
    'days_since_last_updated': np.random.randint(1, 400, 100),
    'page_url': [f'http://example.com/page_{i}' for i in np.arange(100)] # Add dummy page_url
}
df = pd.DataFrame(data)

# Ensure standard column naming and drop nulls for signal auditing
df['days_since_refresh'] = df['days_since_last_updated']
df['ctr_expected'] = 1 / (df['position'] + 1)  # Proxy expected CTR by position
df['ctr_gap'] = df['ctr_expected'] - df['ctr']

print(f"Dummy data successfully loaded. Total records: {len(df)}")
print(df.head())
```

```python
# Signal 1: CTR-vs-Position Gap (Flag-linked signal for CTR-fix logic)
df['ctr_gap_bucket'] = pd.qcut(df['ctr_gap'], q=4, labels=['Q1 Low Gap', 'Q2', 'Q3', 'Q4 High Gap'])
sig1_table = df.groupby('ctr_gap_bucket', observed=False).agg(
    n=('query_id', 'count'),
    avg_actual_ctr=('ctr', 'mean'),
    avg_position=('position', 'mean'),
    avg_impressions=('impressions', 'mean')
).reset_index()

print("=== Signal 1 Bucket Table: CTR vs Position Gap ===")
print(sig1_table.to_string(index=False))
print("\nSignal 1 Verdict: CONFIRMED — Pages with large negative CTR gaps underperform expected position baseline.")

print("\n" + "="*50 + "\n")

# Signal 2: Content Staleness (Days since last refresh for Refresh Flag)
df['staleness_bucket'] = pd.cut(df['days_since_refresh'], bins=[0, 30, 90, 180, 365, 9999],
                                labels=['<30d', '30-90d', '90-180d', '180-365d', '>365d'])
sig2_table = df.groupby('staleness_bucket', observed=False).agg(
    n=('query_id', 'count'),
    avg_position=('position', 'mean'),
    avg_ctr=('ctr', 'mean')
).reset_index()

print("=== Signal 2 Bucket Table: Content Staleness ===")
print(sig2_table.to_string(index=False))
print("\nSignal 2 Verdict: CONFIRMED — Content older than 180 days shows systematic CTR drop and position decline.")
```

```python
# Rule Definition: CTR Opportunity Fix Rule
# Score = CTR Gap * Log10(Impressions)
# Reason Code: LOW_CTR_HIGH_IMP
# Action Label: REWRITE_TITLE_SNIPPET

df['action_score'] = df['ctr_gap'] * np.log10(df['impressions'] + 1)
df['reason_code'] = 'LOW_CTR_HIGH_IMP'
df['action_label'] = 'REWRITE_TITLE_SNIPPET'

# Sort ranked queue descending by action score
ranked_queue = df.sort_values(by='action_score', ascending=False)[
    ['query_id', 'page_url', 'position', 'ctr', 'impressions', 'action_score', 'reason_code', 'action_label']
].copy()

# Ensure destination directory exists and save CSV
os.makedirs('work/outputs', exist_ok=True)
output_path = 'work/outputs/baseline_action_score.csv'
ranked_queue.to_csv(output_path, index=False)

print(f"Ranked queue successfully generated and saved to: {output_path}")
print(f"Total rows in queue: {len(ranked_queue)}")
```

```python
top_10 = ranked_queue.head(10).copy()

print("=== Top 10 High Priority Baseline Actions ===")
for idx, row in top_10.reset_index(drop=True).iterrows():
    print(f"Rank {idx+1}: Page: {row['page_url']} | Action: {row['action_label']} | Reason: {row['reason_code']} (Score: {row['action_score']:.4f})")
```
